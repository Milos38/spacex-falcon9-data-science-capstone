## Imports


```python
import requests
import pandas as pd
import numpy as np
```

## Collect data


```python
url='https://api.spacexdata.com/v4/launches/past'
r=requests.get(url, timeout=30)
r.raise_for_status()
raw=r.json()
df=pd.json_normalize(raw)
print(df.shape)
display(df.head())
```

## Export


```python
df.to_csv("dataset_part_1.csv", index=False)
```

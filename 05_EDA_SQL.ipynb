## Load to SQLite


```python
import pandas as pd
import sqlite3
df=pd.read_csv("dataset_part_2.csv")
con=sqlite3.connect(":memory:")
df.to_sql("SPACEXTBL",con,index=False,if_exists="replace")
```

## Inspect schema


```python
display(pd.read_sql_query('PRAGMA table_info(SPACEXTBL)', con))
```

## Record count


```python
display(pd.read_sql_query('SELECT COUNT(*) AS rows FROM SPACEXTBL', con))
```

## Libraries


```python
import pandas as pd
import folium
```

## Load geographic data


```python
geo=pd.read_csv('spacex_launch_geo.csv')
display(geo.head())
```

## Create map


```python
m=folium.Map(location=[29.5,-95], zoom_start=4)
lat=[c for c in geo.columns if 'lat' in c.lower()]
lon=[c for c in geo.columns if 'lon' in c.lower()]
if lat and lon:
    for _,r in geo.dropna(subset=[lat[0],lon[0]]).iterrows():
        folium.Marker([r[lat[0]],r[lon[0]]]).add_to(m)
m
```
